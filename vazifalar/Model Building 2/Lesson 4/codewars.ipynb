```python
# eng kop like olgan yechim codewarsda
def find_next_square(sq):
    root = sq ** 0.5
    if root.is_integer():            # bu yerda son mukammal kvadratligini tekshirmoqchi bolganimizda bu .is_integer() float son butun son butun son ekanligini aniqlash
        return (root + 1)**2
    return -1
```
