# Data Preprocessing with Class

```python
df.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder,MinMaxScaler
class DataPreprocessing:
    def __init__(self,df):
        self.df=df
    def tozala(self):
        for col in self.df.columns:
            if self.df[col].isnull().any():
                if self.df[col].dtype=='object':
                    self.df[col].fillna(self.df[col].mode()[0],inplace=True)                
                else:
                    self.df[col].fillna(self.df[col].mean(),inplace=True)
        return self
    def encodla(self):
        encoder=LabelEncoder()
        for col in self.df.columns:
            if self.df[col].dtype=='object':
                if self.df[col].nunique()<=5:
                    dummies=pd.get_dummies(self.df[col],prefix=col,dtype=int)
                    self.df=pd.concat([self.df.drop(columns=col),dummies],axis=1)
                else:
                    self.df[col]=encoder.fit_transform(self.df[col])
        return self
    def scale_qil(self):
        scaler=MinMaxScaler()
        for col in self.df.columns:
            num_col=self.df.select_dtypes(include=['int64','float64']).columns.drop('telefon_turi')
            self.df[num_col]=scaler.fit_transform(self.df[num_col])
        return self
            
```

```python
dp=DataPreprocessing(df)
dp.tozala().encodla().scale_qil()
preprocessed_data=dp.df
```

# Model Training Process

```python
# separating input and output features,respectively  x and y. 
from sklearn.model_selection import train_test_split
```

```python
x=df.drop('telefon_turi',axis=1)
y=df['telefon_turi']
```

```python
# The dataset is divided into two separate parts: a training  and a testing 
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

# Model Selection

# Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier()
```

```python
rf.fit(x_train,y_train)
```

```python
y_pred=rf.predict(x_test)
```

```python
from sklearn.metrics import classification_report,accuracy_score
```

```python
report_rf=classification_report(y_test,y_pred)
```
