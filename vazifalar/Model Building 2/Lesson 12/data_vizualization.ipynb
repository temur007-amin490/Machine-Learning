# Loding Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Uplodading data

```python
df=pd.read_csv('cars2018.csv')
```

# Introducing with Data

```python
df.info()
```

```python
df.head()
```

```python
df = df.drop('Unnamed: 15', axis=1)
```

```python
import klib
```

```python
klib.missingval_plot(df)
```

```python
# bar chart
plt.figure(figsize=(10,6))
plt.bar(df['aspiration'], df['cylinders'], linestyle='-', color='blue', label='Sales')
plt.title('importanse of cylinders in Engine aspiration ', fontsize=16)
plt.xlabel('Engine aspiration', fontsize=12)
plt.ylabel('cylinders', fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.boxplot(x='mpg',y='recommended_fuel',data=df)
plt.title('The importance of classes to recommended_fuel')
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.countplot(x='transmission',hue='drive',data=df,palette='Set1')
plt.title('the importance of transmission in car operating')                    # 
plt.show
```

```python
plt.figure(figsize=(10,4))
sns.histplot(data=df, x='aspiration', hue='recommended_fuel', multiple='stack', bins=30)
plt.title('model vs fuel_injection')
plt.show()
```
