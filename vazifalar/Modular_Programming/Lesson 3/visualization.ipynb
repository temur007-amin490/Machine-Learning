```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('Tech_Use_Stress_Wellness.csv')
```

```python
df.info()
```

```python
df.head()
```

```python
import klib
```

```python
klib.missingval_plot(df)
```

```python
# bar chart
plt.figure(figsize=(10,6))
plt.bar(df['phone_usage_hours'], df['mental_health_score'], linestyle='-', color='blue', label='Sales')
plt.title('importanse of phone usage hours in mental health score ', fontsize=16)
plt.xlabel('phone_usage_hours', fontsize=12)
plt.ylabel('mental_health_score', fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.boxplot(x='mindfulness_minutes_per_day',y='physical_activity_hours_per_week',data=df)
plt.title('The importance of physical activity in mindfulness')
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.countplot(x='caffeine_intake_mg_per_day',hue='gender',data=df,palette='Set1')
plt.title('the importance of caffeine in gender')                    # 
plt.show
```

```python
plt.figure(figsize=(10,4))
sns.histplot(data=df, x='social_media_hours', hue='work_related_hours', multiple='stack', bins=30)
plt.title('social media vs work life')
plt.show()
```
